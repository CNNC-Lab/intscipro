```python
from IPython.display import HTML, display
```

```python
data = [
    {'subject_id': 1, 'age': 25, 'gender': 'M', 'score': 88},
    {'subject_id': 2, 'age': 30, 'gender': 'F', 'score': 92},
    {'subject_id': 3, 'age': 35, 'gender': 'M', 'score': 87},
]
```

This is a nice table with some results

```python
header = ''.join(f'<th>{key}</th>' for key in data[0])
rows = ''.join(
    '<tr>' + ''.join(f'<td>{value}</td>' for value in row.values()) + '</tr>'
    for row in data
)
display(HTML(f'<table><tr>{header}</tr>{rows}</table>'))

for row in data:
    print(row)
```
